### Тест на 4 монетах (близко)

```python
def coin_classification1_1(image_path, GaussMatrix, dp, minDist,
                         param1, param2, minRadius, maxRadius):
  #1 Сбор изображения
  image = cv.imread(image_path)


  #2 Предварительная обработка изображения
      # Преобразуйте изображение в оттенки серого
  gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
      # Примените размытие Гаусса для уменьшения шума
  gray = cv.GaussianBlur(gray, GaussMatrix, 0)
      # Используйте оператор Кэнни для обнаружения краев
  edges = cv.Canny(gray, 100, 200, apertureSize=3)


  #3 Обнаружение кругов
      # Используйте преобразование Хафа для обнаружения кругов на изображении
  dp = dp          # Параметр разрешения аккумуляторного массива
  minDist = minDist     # Минимальное расстояние между центрами
  param1 = param1       # Порог для метода Кэнни
  param2 = param2      # Порог для функции HoughCircles (чем меньше, тем больше ложных кругов)
  minRadius = minRadius    # Минимальный радиус круга
  maxRadius = maxRadius   # Максимальный радиус круга

  circles = cv.HoughCircles(gray, cv.HOUGH_GRADIENT, dp, minDist,
                            param1=param1, param2=param2,
                            minRadius=minRadius, maxRadius=maxRadius)
      # Получите координаты центров и радиусов обнаруженных кругов
  img_circles = image.copy()

  if circles is not None:
    circles = np.uint16(np.around(circles))
    max_radius = max([c[2] for c in circles[0, :]])
    coin_summ = 0

    for circle in circles[0, :]:
      center = (circle[0], circle[1])
      radius = circle[2]

      cv.circle(img_circles, center, radius, (0, 0, 255), 8)
      cv.circle(img_circles, center, 2, (0, 0, 255), 9)

      if radius < int(max_radius * 0.85):
        coin_value = '1'
      elif radius < int(max_radius * 0.9):
        coin_value = '10'
      elif radius < int(max_radius * 0.95):
        coin_value = '2'
      else:
        coin_value = '5'

      fontFace = cv.FONT_HERSHEY_COMPLEX
      fontScale = 3
      cv.putText(img_circles, coin_value, (circle[0]-50, circle[1]+30), fontFace, fontScale, (0, 0, 255), 1)
      # str(int(radius))
      coin_summ += int(coin_value)
    print(sorted([int(c[2]) for c in circles[0, :]]))

    cv.putText(img_circles, f'Общая сумма: {coin_summ}', (10, img_circles.shape[0]-20), cv.FONT_HERSHEY_COMPLEX, 3, (0,0,0), 1)
  else:
    print('Круги не найдены')

  plt.figure(figsize=(15,10))

  plt.subplot(1,2,1)
  plt.imshow(edges, cmap='gray')

  plt.subplot(1,2,2)
  plt.imshow(cv.cvtColor(img_circles, cv.COLOR_BGR2RGB))

  plt.show()
```

```python
coin_classification1_1('test1_1.jpg', (3,3), 1.2, 300,
                     50, 100, 250, 350)
```

```python
def coin_classification(image_path, GaussMatrix, dp, minDist,
                         param1, param2, minRadius, maxRadius):
  #1 Сбор изображения
  image = cv.imread(image_path)


  #2 Предварительная обработка изображения
      # Преобразуйте изображение в оттенки серого
  gray = cv.cvtColor(image, cv.COLOR_BGR2GRAY)
      # Примените размытие Гаусса для уменьшения шума
  gray = cv.GaussianBlur(gray, GaussMatrix, 0)
      # Используйте оператор Кэнни для обнаружения краев
  edges = cv.Canny(gray, 100, 200, apertureSize=3)


  #3 Обнаружение кругов
      # Используйте преобразование Хафа для обнаружения кругов на изображении
  dp = dp          # Параметр разрешения аккумуляторного массива
  minDist = minDist     # Минимальное расстояние между центрами
  param1 = param1       # Порог для метода Кэнни
  param2 = param2      # Порог для функции HoughCircles (чем меньше, тем больше ложных кругов)
  minRadius = minRadius    # Минимальный радиус круга
  maxRadius = maxRadius   # Максимальный радиус круга

  circles = cv.HoughCircles(gray, cv.HOUGH_GRADIENT, dp, minDist,
                            param1=param1, param2=param2,
                            minRadius=minRadius, maxRadius=maxRadius)
      # Получите координаты центров и радиусов обнаруженных кругов
  img_circles = image.copy()

  if circles is not None:
    circles = np.uint16(np.around(circles))
    max_radius = max([c[2] for c in circles[0, :]])
    coin_summ = 0

    r1 = int(max_radius * 0.85)
    r10 = int(max_radius * 0.9)
    r2 = int(max_radius * 0.925)

    if max_radius > 290:
      r2 = int(max_radius * 0.95)
    if (max_radius > 135) and (max_radius < 155):
      r10 = int(max_radius * 0.88)

    for circle in circles[0, :]:
      center = (circle[0], circle[1])
      radius = circle[2]

      cv.circle(img_circles, center, radius, (0, 0, 255), 9)
      cv.circle(img_circles, center, 2, (0, 0, 255), 9)

      if radius < r1:
        coin_value = '1'
      elif radius < r10:
        coin_value = '10'
      elif radius < r2:
        coin_value = '2'
      else:
        coin_value = '5'

      fontFace = cv.FONT_HERSHEY_COMPLEX
      fontScale = 3
      cv.putText(img_circles, coin_value, (circle[0]-45, circle[1]+30), fontFace, fontScale, (0, 0, 255), 1)

      coin_summ += int(coin_value)
    # print(sorted([int(c[2]) for c in circles[0, :]]))

    cv.putText(img_circles, f'Общая сумма: {coin_summ}', (10, img_circles.shape[0]-20), cv.FONT_HERSHEY_COMPLEX, 3, (0,0,0), 1)
  else:
    print('Круги не найдены')

  return cv.cvtColor(img_circles, cv.COLOR_BGR2RGB)
```
